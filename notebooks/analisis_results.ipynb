```python
# 1. Definimos la lista de columnas que nos interesan para TinderJob
columnas_seleccionadas = [
    'ResponseId', 'DevType', 'Country', 'ConvertedCompYearly', 
    'WorkExp', 'YearsCode', 'Industry', 'OrgSize', 'Employment',
    'LanguageHaveWorkedWith', 'DatabaseHaveWorkedWith', 'Age', 'EdLevel'
]

# 2. Creamos el sub-dataset de trabajo filtrando las columnas
df_tinder_stackoverflow = df_results_original[columnas_seleccionadas].copy()

# 3. Filtramos inmediatamente para quedarnos solo con los profesionales en España
df_tinder_spain = df_tinder_stackoverflow[df_tinder_stackoverflow['Country'] == 'Spain'].copy()

# 4. Echamos un vistazo al nuevo tamaño reducido y limpio
print("Dimensiones del dataset de España:", df_tinder_spain.shape)
df_tinder_spain.head()
```
