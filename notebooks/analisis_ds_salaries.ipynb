### `df.head()` - Sacamos las primeras líneas para ver su estructura general

### `df.shape` - Dimensiones del dataset (tupla)

### `df.describe()` - Estadisticas descriptivas del Dataset

```python
print("\n--- Estadísticas descriptivas ---")
print()
print(df_ds_salaries_original.describe())
```

```python
df_euros = df_ds_salaries_original[df_ds_salaries_original['salary_currency'] == 'EUR'] # Filtra el DataFrame para obtener solo los registros con salario en euros
print("\n--- Estadísticas descriptivas para salarios en euros ---")
print()
print(df_euros.describe())
print(f"\nTendríamos ", df_euros.shape[0], " registros con salario en euros.")
```

## CREACIÓN COPIA DE ds_salaries.csv

```python
df_limpio = df_ds_salaries_original.copy()  # Realiza una copia del DataFrame original para trabajar con él sin modificar el original
```

### Creamos columna EUR en base a los salarios de USD

```python
df_limpio['salary_in_eur'] = df_limpio['salary_in_usd'] * 0.92  # Crea una nueva columna 'salary_in_eur' convirtiendo los salarios de USD a EUR usando un tipo de cambio aproximado
print("\n--- Estadísticas descriptivas para la nueva columna 'salary_in_eur' ---")
print() 
print(df_limpio['salary_in_eur'].describe())
print(df_limpio.describe())
```

```python
df_spain = df_limpio[df_limpio['company_location'] == 'ES'].copy() # Filtra el DataFrame para obtener solo los registros con ubicación en España y crea una copia para trabajar con ella sin modificar el DataFrame original
print("Tendríamos ", df_spain.shape[0], " registros con ubicación en España.")
print("\n--- Estadísticas descriptivas para salarios en España ---")
print()
print(df_spain['salary_in_eur'].describe())
print(df_spain.describe())
print()
```
